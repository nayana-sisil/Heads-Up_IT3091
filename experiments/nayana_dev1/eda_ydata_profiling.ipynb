**Note on dataset size:** this dataset has ~180,000 rows and 50+ columns. The full profiling mode can be slow and memory-heavy at this scale, so this notebook uses `minimal=True` by default, which skips the most expensive pairwise-correlation and interaction computations. Set `minimal=False` below if you want the full report and have the time/memory to spare (can take several minutes).


````python
import pandas as pd
from pathlib import Path
from ydata_profiling import ProfileReport

DATA_PATH = Path('../../data/raw data/DataCoSupplyChainDataset.csv')  
OUTPUT_PATH = Path('../../experiments/nayana_dev1/profiling_report.html')

MINIMAL_MODE = True  # set to False for the full report (slower, more detailed)
````

````python
import sys
print(sys.executable)
````

## 1. Load the data




````python
try:
    df = pd.read_csv(DATA_PATH, encoding='utf-8')
except UnicodeDecodeError:
    print("UTF-8 failed, falling back to ISO-8859-1")
    df = pd.read_csv(DATA_PATH, encoding='ISO-8859-1')

print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
````

## 2. Generate the profiling report

This single call produces the full report object: per-column type inference, distributions, missing-value analysis, duplicate rows, and (in full mode) correlation matrices and variable interactions.


````python
profile = ProfileReport(
    df,
    title="Heads-Up — DataCo Smart Supply Chain Profiling Report",
    minimal=MINIMAL_MODE,
    explorative=not MINIMAL_MODE,
)
````

## 3. Save as a standalone HTML file (don't rely on inline rendering)



````python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
profile.to_file(OUTPUT_PATH)
print(f"Report saved to: {OUTPUT_PATH.resolve()}")
````
